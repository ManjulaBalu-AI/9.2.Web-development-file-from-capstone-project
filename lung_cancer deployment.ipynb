```python
import pickle
```

```python
# Load the saved model
filename = "finalized_model_log_reg_selected_5_features.sav"
```

```python
with open(filename, "rb") as f:
    model = pickle.load(f)
```

```python
print("Model loaded successfully!")
print("Model expects", model.n_features_in_, "features")

# Enter the 5 selected features
data1 = int(input("ALLERGY (1 or 2): "))
data2 = int(input("WHEEZING (1 or 2): "))
data3 = int(input("ALCOHOL CONSUMING(1 or 2): "))
data4 = int(input("COUGHING (1 or 2): "))
data5 = int(input("SWALLOWING DIFFICULTY(1 or 2): "))

# Create input
input_data = [[
    data1,
    data2,
    data3,
    data4,
    data5
]]

# Prediction
prediction = model.predict(input_data)

print("\nPrediction:", prediction[0])

if prediction[0] == 1:
    print("Result: Lung Cancer Positive")
else:
    print("Result: Lung Cancer Negative")
```
